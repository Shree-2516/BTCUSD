CELL 1 — Imports

```python
import pandas as pd
import numpy as np
import talib
```

CELL 2 — Load CSV Data

```python
DATA_PATH = "../data/btcusdt_5m_2020_2026.csv"

df = pd.read_csv(DATA_PATH)

df.head()
```

CELL 3 — Convert Datetime

```python
df["datetime"] = pd.to_datetime(df["datetime"])

df.set_index("datetime", inplace=True)

df.head()
```

CELL 4 — Check Dataset Shape

```python
print(df.shape)

df.tail()
```

CELL 5 — RSI Indicator

RSI = momentum indicator.

```python
df["rsi_14"] = talib.RSI(
    df["close"],
    timeperiod=14
)

df[["close", "rsi_14"]].tail()
```

CELL 6 — EMA Indicators

```python
df["ema_20"] = talib.EMA(
    df["close"],
    timeperiod=20
)

df["ema_50"] = talib.EMA(
    df["close"],
    timeperiod=50
)

df["ema_200"] = talib.EMA(
    df["close"],
    timeperiod=200
)

df[
    [
        "close",
        "ema_20",
        "ema_50",
        "ema_200"
    ]
].tail()
```

CELL 7 — ATR (Volatility)

ATR = volatility/range indicator.

```python
df["atr_14"] = talib.ATR(
    df["high"],
    df["low"],
    df["close"],
    timeperiod=14
)

df[["close", "atr_14"]].tail()
```

CELL 8 — MACD

```python
macd, macd_signal, macd_hist = talib.MACD(
    df["close"],
    fastperiod=12,
    slowperiod=26,
    signalperiod=9
)

df["macd"] = macd
df["macd_signal"] = macd_signal
df["macd_hist"] = macd_hist

df[
    [
        "macd",
        "macd_signal",
        "macd_hist"
    ]
].tail()
```

CELL 9 — Bollinger Bands

```python
upper, middle, lower = talib.BBANDS(
    df["close"],
    timeperiod=20
)

df["bb_upper"] = upper
df["bb_middle"] = middle
df["bb_lower"] = lower

df[
    [
        "close",
        "bb_upper",
        "bb_lower"
    ]
].tail()
```

CELL 10 — Candle Body & Wick Features

VERY important for price action AI.

```python
df["body_size"] = abs(
    df["close"] - df["open"]
)

df["upper_wick"] = (
    df["high"] - df[["open", "close"]].max(axis=1)
)

df["lower_wick"] = (
    df[["open", "close"]].min(axis=1) - df["low"]
)

df["candle_range"] = (
    df["high"] - df["low"]
)

df[
    [
        "body_size",
        "upper_wick",
        "lower_wick",
        "candle_range"
    ]
].tail()
```

CELL 11 — Bullish / Bearish Candle

```python
df["bullish_candle"] = (
    df["close"] > df["open"]
).astype(int)

df["bearish_candle"] = (
    df["close"] < df["open"]
).astype(int)

df[
    [
        "open",
        "close",
        "bullish_candle",
        "bearish_candle"
    ]
].tail()
```

CELL 12 — Engulfing Pattern

```python
df["engulfing"] = talib.CDLENGULFING(
    df["open"],
    df["high"],
    df["low"],
    df["close"]
)

df[
    [
        "open",
        "high",
        "low",
        "close",
        "engulfing"
    ]
].tail(20)
```

CELL 13 — Doji Pattern

```python
df["doji"] = talib.CDLDOJI(
    df["open"],
    df["high"],
    df["low"],
    df["close"]
)

df[
    [
        "open",
        "close",
        "doji"
    ]
].tail(20)
```

CELL 14 — Hammer Pattern

```python
df["hammer"] = talib.CDLHAMMER(
    df["open"],
    df["high"],
    df["low"],
    df["close"]
)

df[
    [
        "open",
        "close",
        "hammer"
    ]
].tail(20)
```

CELL 15 — Trend Detection

Simple EMA trend logic.

```python
df["trend"] = np.where(
    df["ema_20"] > df["ema_50"],
    "bullish",
    "bearish"
)

df[
    [
        "close",
        "ema_20",
        "ema_50",
        "trend"
    ]
].tail()
```

CELL 16 — Support & Resistance Approximation

Rolling highs/lows.

```python
df["resistance"] = (
    df["high"]
    .rolling(50)
    .max()
)

df["support"] = (
    df["low"]
    .rolling(50)
    .min()
)

df[
    [
        "close",
        "support",
        "resistance"
    ]
].tail()
```

CELL 17 — Future Target Label

VERY IMPORTANT for AI training.

Predict:

next candle bullish or bearish

```python
df["future_close"] = df["close"].shift(-1)

df["target"] = np.where(
    df["future_close"] > df["close"],
    1,
    0
)

df[
    [
        "close",
        "future_close",
        "target"
    ]
].tail()
```

CELL 18 — Remove NaN Rows

Indicators create NaN initially.

```python
df.dropna(inplace=True)

print(df.shape)
```

CELL 19 — Save Feature Dataset

```python
SAVE_PATH = "../data/btcusdt_features.parquet"

df.to_parquet(SAVE_PATH)

print("Saved:", SAVE_PATH)
```
